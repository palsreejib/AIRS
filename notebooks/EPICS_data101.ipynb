## Folder Formatting

```python
import os
import shutil
import random
from pathlib import Path

# Set your paths
source_dir = r"D:\DOCUMENTS\IQUISITIONIS\204_EPICS\ROSE-LEAF-DISEASES-SYMPTOMS-ID-AND-CONTROL-GUIDE\A COMPREHENSIVE DATASET ON DISEASES AFFECTING ROSE LEAVES IDENTIFICATION, SYMPTOMS, AND CONTROL STRATEGIES\ROSE LEAF\Rose Leaf"
target_dir = r"D:\DOCUMENTS\IQUISITIONIS\203_VITFACIALRECC\DATA\ROSE-LEAF-DISEASE-DATASET"
test_ratio = 0.2  # 20% for test, 80% for train

# Create target directory structure
os.makedirs(os.path.join(target_dir, "test"), exist_ok=True)
os.makedirs(os.path.join(target_dir, "train"), exist_ok=True)

# Process each disease/health category folder
for category in os.listdir(source_dir):
    category_path = os.path.join(source_dir, category)
    if not os.path.isdir(category_path):
        continue
    
    # Create train and test directories for this category
    train_dir = os.path.join(target_dir, "train", category)
    test_dir = os.path.join(target_dir, "test", category)
    os.makedirs(train_dir, exist_ok=True)
    os.makedirs(test_dir, exist_ok=True)
    
    # Get all image files for this category
    images = [f for f in os.listdir(category_path) 
              if os.path.isfile(os.path.join(category_path, f))]
    
    # Shuffle and split (80/20)
    random.shuffle(images)
    split_idx = int(len(images) * test_ratio)
    test_images = images[:split_idx]
    train_images = images[split_idx:]
    
    # Copy files to their new locations
    for img in test_images:
        src = os.path.join(category_path, img)
        dst = os.path.join(test_dir, img)
        shutil.copy2(src, dst)
    
    for img in train_images:
        src = os.path.join(category_path, img)
        dst = os.path.join(train_dir, img)
        shutil.copy2(src, dst)
    
    print(f"Processed {category}: {len(train_images)} train, {len(test_images)} test images")

print("Dataset reorganization complete!")
print(f"New dataset structure created at: {target_dir}")
```
